```python
import matplotlib.pyplot as plt
import numpy as np

# Model names
models = ['LightGCN', 'GraphSAGE', 'GAT', 'PinSAGE']
x = np.arange(len(models))

# Data (from your result table)
train_auc  = [0.9094, 0.9894, 0.9719, 0.9298]
val_auc    = [0.8355, 0.9860, 0.9729, 0.9562]

train_ap   = [0.9197, 0.9848, 0.9945, 0.9434]
val_ap     = [0.8592, 0.9798, 0.9888, 0.9649]

epoch_time = [569.17, 16.19, 2.55, 1881.0]

# Setup: 1 row, 3 columns
fig, axes = plt.subplots(1, 3, figsize=(10, 4), sharex=False)

# ---- AUC Plot ----
axes[0].scatter(models, train_auc, marker='D', label='Train AUC')
axes[0].scatter(models, val_auc, marker='D', label='Val AUC')
axes[0].set_ylim(0.8, 1.0)
axes[0].set_title('AUC')
axes[0].set_ylabel('Score')
axes[0].grid(True, axis='y', linestyle='--', alpha=0.4)
axes[0].legend()

# ---- AP Plot ----
axes[1].scatter(models, train_ap, marker='D', label='Train AP')
axes[1].scatter(models, val_ap, marker='D', label='Val AP')
axes[1].set_ylim(0.8, 1.0)
axes[1].set_title('Average Precision')
axes[1].grid(True, axis='y', linestyle='--', alpha=0.4)
axes[1].legend()

# ---- Epoch Time Plot ----
axes[2].scatter(models, epoch_time, marker='D', color='gray', label='Epoch Time')
axes[2].set_yscale('log')
axes[2].set_title('Epoch Time (s)')
axes[2].grid(True, axis='y', linestyle='--', alpha=0.4)
axes[2].legend()

plt.rcParams.update({'font.size': 14})
axes[0].tick_params(axis='x', labelrotation=15)
axes[1].tick_params(axis='x', labelrotation=15)
axes[2].tick_params(axis='x', labelrotation=15)


# Tidy layout
plt.tight_layout()
plt.show()


```
